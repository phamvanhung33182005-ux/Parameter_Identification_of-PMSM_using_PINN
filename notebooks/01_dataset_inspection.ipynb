```python
# ===== CELL 1: KẾT NỐI GOOGLE DRIVE VÀ XÁC ĐỊNH DATASET =====

from google.colab import drive  # Import công cụ để kết nối Google Drive với Google Colab.
import os  # Import thư viện os để thao tác với thư mục và đường dẫn.
import pandas as pd  # Import pandas để đọc và xử lý dữ liệu CSV.

drive.mount('/content/drive')  # Gắn Google Drive vào môi trường Colab.

DRIVE_PROJECT = '/content/drive/MyDrive/Parameter_Identification_of_PMSM_using_PINN'  # Khai báo đường dẫn project trên Google Drive.

RAW_DATA_DIR = os.path.join(DRIVE_PROJECT, 'dataset', 'raw')  # Tạo đường dẫn đến thư mục chứa 5 file CSV gốc.

DATA_FILES = ['Data1.csv', 'Data2.csv', 'Data3.csv', 'Data4.csv', 'Data5.csv']  # Khai báo chính xác tên 5 dataset.

print('Project tồn tại:', os.path.exists(DRIVE_PROJECT))  # Kiểm tra thư mục project có tồn tại hay không.

print('Raw dataset tồn tại:', os.path.exists(RAW_DATA_DIR))  # Kiểm tra thư mục raw có tồn tại hay không.

if os.path.exists(RAW_DATA_DIR):  # Chỉ kiểm tra các file nếu thư mục raw thực sự tồn tại.
    print('Các file trong raw:', os.listdir(RAW_DATA_DIR))  # In danh sách file hiện có trong thư mục raw.
```

```python
# ===== CELL 2: ĐỌC 5 DATASET VÀ KIỂM TRA KÍCH THƯỚC =====

datasets = {}  # Tạo dictionary để lưu DataFrame của từng dataset.

for file_name in DATA_FILES:  # Lặp qua từng file trong danh sách 5 dataset.
    file_path = os.path.join(RAW_DATA_DIR, file_name)  # Tạo đường dẫn đầy đủ đến file CSV hiện tại.
    datasets[file_name] = pd.read_csv(file_path)  # Đọc file CSV và lưu DataFrame vào dictionary.
    print(f'{file_name}: {datasets[file_name].shape}')  # In số dòng và số cột của dataset vừa đọc.
```

```python
# ===== CELL 3: KIỂM TRA CẤU TRÚC VÀ CHẤT LƯỢNG 5 DATASET =====

print('===== KIỂM TRA CẤU TRÚC DATASET =====')  # In tiêu đề phần kiểm tra cấu trúc dữ liệu.

reference_columns = list(datasets[DATA_FILES[0]].columns)  # Lấy danh sách cột của Data1.csv làm cấu trúc tham chiếu.

print('Số cột tham chiếu:', len(reference_columns))  # In số lượng cột của Data1.csv.

print('Tên các cột:', reference_columns)  # In toàn bộ tên các cột của Data1.csv.

for file_name, df in datasets.items():  # Lặp qua từng dataset trong dictionary.
    same_columns = list(df.columns) == reference_columns  # Kiểm tra dataset hiện tại có cùng tên và thứ tự cột với Data1.csv hay không.
    print(f'{file_name}: shape={df.shape}, cùng cấu trúc={same_columns}')  # In kích thước và kết quả kiểm tra cấu trúc của từng dataset.

print('\n===== KIỂM TRA GIÁ TRỊ THIẾU =====')  # In tiêu đề phần kiểm tra giá trị thiếu.

for file_name, df in datasets.items():  # Lặp qua từng dataset để kiểm tra NaN.
    missing_count = df.isna().sum().sum()  # Đếm tổng số giá trị NaN trong toàn bộ dataset.
    print(f'{file_name}: số giá trị thiếu = {missing_count}')  # In số lượng giá trị thiếu của từng dataset.

print('\n===== KIỂM TRA GIÁ TRỊ VÔ HẠN =====')  # In tiêu đề phần kiểm tra giá trị vô hạn.

for file_name, df in datasets.items():  # Lặp qua từng dataset để kiểm tra inf và -inf.
    numeric_df = df.select_dtypes(include=np.number)  # Lấy các cột có kiểu dữ liệu số để kiểm tra.
    inf_count = np.isinf(numeric_df.to_numpy()).sum()  # Đếm tổng số giá trị vô hạn trong các cột số.
    print(f'{file_name}: số giá trị vô hạn = {inf_count}')  # In số lượng giá trị vô hạn của từng dataset.

print('\n===== KIỂM TRA KIỂU DỮ LIỆU =====')  # In tiêu đề phần kiểm tra kiểu dữ liệu.

for file_name, df in datasets.items():  # Lặp qua từng dataset để kiểm tra kiểu dữ liệu.
    print(f'\n{file_name}:')  # In tên dataset hiện tại.
    print(df.dtypes)  # In kiểu dữ liệu của từng cột trong dataset.
```

```python
# ===== CELL 4: THỐNG KÊ PHẠM VI INPUT VÀ TARGET =====

INPUT_COLUMNS = ['ud', 'uq', 'Id', 'Iq', 'nm']  # Khai báo 5 biến đầu vào sẽ được sử dụng cho PINN.

TARGET_COLUMNS = ['Rs', 'Ld', 'Lq']  # Khai báo 3 tham số động cơ cần nhận dạng.

print('===== THỐNG KÊ INPUT =====')  # In tiêu đề phần thống kê các biến đầu vào.

for column in INPUT_COLUMNS:  # Lặp qua từng biến đầu vào cần kiểm tra.
    print(f'\n--- {column} ---')  # In tên biến đang được thống kê.
    print(pd.concat([df[column].describe() for df in datasets.values()], axis=1, keys=datasets.keys()))  # In các thống kê count, mean, std, min, quartile và max của biến trên 5 dataset.

print('\n===== THỐNG KÊ TARGET =====')  # In tiêu đề phần thống kê các tham số cần nhận dạng.

for column in TARGET_COLUMNS:  # Lặp qua từng tham số Rs, Ld và Lq.
    print(f'\n--- {column} ---')  # In tên tham số đang được thống kê.
    print(pd.concat([df[column].describe() for df in datasets.values()], axis=1, keys=datasets.keys()))  # In thống kê của tham số trên 5 dataset.
```

```python
# ===== CELL 5: PHÂN TÍCH MỐI QUAN HỆ GIỮA INPUT VÀ TARGET =====

ANALYSIS_COLUMNS = INPUT_COLUMNS + TARGET_COLUMNS  # Ghép 5 biến input và 3 biến target thành danh sách cần phân tích.

print('===== MA TRẬN TƯƠNG QUAN =====')  # In tiêu đề phần phân tích tương quan.

for file_name, df in datasets.items():  # Lặp qua từng dataset để phân tích riêng từng file.
    print(f'\n===== {file_name} =====')  # In tên dataset đang được phân tích.
    correlation = df[ANALYSIS_COLUMNS].corr()  # Tính hệ số tương quan Pearson giữa các biến được chọn.
    print(correlation.round(3))  # In ma trận tương quan và làm tròn 3 chữ số thập phân.

print('\n===== ĐỘ BIẾN THIÊN CỦA INPUT =====')  # In tiêu đề phần kiểm tra độ biến thiên của input.

for column in INPUT_COLUMNS:  # Lặp qua từng biến đầu vào.
    minimum = min(df[column].min() for df in datasets.values())  # Tìm giá trị nhỏ nhất của biến trên toàn bộ 5 dataset.
    maximum = max(df[column].max() for df in datasets.values())  # Tìm giá trị lớn nhất của biến trên toàn bộ 5 dataset.
    mean = np.mean([df[column].mean() for df in datasets.values()])  # Tính trung bình các giá trị trung bình của 5 dataset.
    print(f'{column}: min={minimum:.6g}, max={maximum:.6g}, mean≈{mean:.6g}')  # In phạm vi và giá trị trung bình của biến.

print('\n===== ĐỘ BIẾN THIÊN CỦA TARGET =====')  # In tiêu đề phần kiểm tra độ biến thiên của target.

for column in TARGET_COLUMNS:  # Lặp qua từng tham số Rs, Ld và Lq.
    minimum = min(df[column].min() for df in datasets.values())  # Tìm giá trị nhỏ nhất của tham số trên toàn bộ 5 dataset.
    maximum = max(df[column].max() for df in datasets.values())  # Tìm giá trị lớn nhất của tham số trên toàn bộ 5 dataset.
    mean = np.mean([df[column].mean() for df in datasets.values()])  # Tính trung bình các giá trị trung bình của 5 dataset.
    print(f'{column}: min={minimum:.6g}, max={maximum:.6g}, mean≈{mean:.6g}')  # In phạm vi và giá trị trung bình của tham số.
```

```python
# ===== CELL 6: KIỂM TRA OPERATING POINT VÀ PHÂN BỐ THEO THỜI GIAN =====

CHECK_COLUMNS = ['Time', 'nm', 'ud', 'uq', 'Id', 'Iq', 'Rs', 'Ld', 'Lq']  # Khai báo các biến cần kiểm tra theo thời gian và operating point.

for file_name, df in datasets.items():  # Lặp qua từng dataset trong 5 dataset.
    print(f'\n===== {file_name} =====')  # In tên dataset đang được kiểm tra.
    print(df[CHECK_COLUMNS].iloc[::5000].to_string(index=False))  # Lấy mỗi 5000 mẫu một điểm để quan sát dữ liệu tại các thời điểm khác nhau.

print('\n===== PHẠM VI OPERATING POINT CỦA 5 DATASET =====')  # In tiêu đề phần tổng hợp operating point.

summary = []  # Tạo danh sách để lưu thông tin tổng hợp của từng dataset.

for file_name, df in datasets.items():  # Lặp qua từng dataset để tính phạm vi các biến.
    summary.append({  # Thêm thông tin thống kê của dataset hiện tại vào danh sách.
        'Dataset': file_name,  # Lưu tên dataset.
        'Time_min': df['Time'].min(),  # Lưu thời điểm nhỏ nhất.
        'Time_max': df['Time'].max(),  # Lưu thời điểm lớn nhất.
        'nm_min': df['nm'].min(),  # Lưu tốc độ nhỏ nhất theo rpm.
        'nm_max': df['nm'].max(),  # Lưu tốc độ lớn nhất theo rpm.
        'Id_min': df['Id'].min(),  # Lưu dòng Id nhỏ nhất.
        'Id_max': df['Id'].max(),  # Lưu dòng Id lớn nhất.
        'Iq_min': df['Iq'].min(),  # Lưu dòng Iq nhỏ nhất.
        'Iq_max': df['Iq'].max(),  # Lưu dòng Iq lớn nhất.
        'Rs_min': df['Rs'].min(),  # Lưu Rs nhỏ nhất.
        'Rs_max': df['Rs'].max(),  # Lưu Rs lớn nhất.
        'Ld_min': df['Ld'].min(),  # Lưu Ld nhỏ nhất.
        'Ld_max': df['Ld'].max(),  # Lưu Ld lớn nhất.
        'Lq_min': df['Lq'].min(),  # Lưu Lq nhỏ nhất.
        'Lq_max': df['Lq'].max()  # Lưu Lq lớn nhất.
    })  # Kết thúc thông tin thống kê của dataset hiện tại.

operating_summary = pd.DataFrame(summary)  # Chuyển danh sách thống kê thành DataFrame để dễ quan sát.

print(operating_summary.round(6).to_string(index=False))  # In bảng tổng hợp và làm tròn các giá trị để dễ đọc.
```
