```python
# ===== CELL 1: KẾT NỐI GOOGLE DRIVE VÀ KIỂM TRA ĐƯỜNG DẪN DATASET =====

from google.colab import drive  # Import công cụ để kết nối Google Drive.
import os  # Import thư viện os để thao tác với đường dẫn.
import pandas as pd  # Import pandas để đọc dữ liệu CSV.
import numpy as np  # Import numpy để kiểm tra dữ liệu số.

drive.mount('/content/drive')  # Kết nối Google Drive với môi trường Google Colab.

DRIVE_PROJECT = '/content/drive/MyDrive/Parameter_Identification_of_PMSM_using_PINN'  # Khai báo thư mục gốc của project trên Drive.

RAW_DATA_DIR = os.path.join(DRIVE_PROJECT, 'dataset', 'raw')  # Tạo đường dẫn đến thư mục chứa dataset gốc.

print('Đường dẫn project:', DRIVE_PROJECT)  # In đường dẫn project để kiểm tra.

print('Đường dẫn raw:', RAW_DATA_DIR)  # In đường dẫn thư mục raw để kiểm tra.

print('Project tồn tại:', os.path.exists(DRIVE_PROJECT))  # Kiểm tra thư mục project có tồn tại hay không.

print('Raw tồn tại:', os.path.exists(RAW_DATA_DIR))  # Kiểm tra thư mục raw có tồn tại hay không.

print('Các file trong raw:', os.listdir(RAW_DATA_DIR))  # In toàn bộ file trong thư mục raw.
```

```python
# ===== CELL 2: ĐỌC TOÀN BỘ 5 DATASET =====

DATA_FILES = ['Data1.csv', 'Data2.csv', 'Data3.csv', 'Data4.csv', 'Data5.csv']  # Khai báo tên chính xác của 5 file CSV.

datasets = {}  # Tạo dictionary để lưu DataFrame của từng dataset.

for file_name in DATA_FILES:  # Lặp qua từng file CSV trong danh sách 5 dataset.
    file_path = os.path.join(RAW_DATA_DIR, file_name)  # Tạo đường dẫn đầy đủ đến file CSV hiện tại.

    print('Đang kiểm tra:', file_path)  # In đường dẫn file hiện tại để biết Colab đang tìm file nào.

    if not os.path.exists(file_path):  # Kiểm tra file CSV hiện tại có thực sự tồn tại hay không.
        raise FileNotFoundError(f'Không tìm thấy file: {file_path}')  # Dừng chương trình và báo chính xác file bị thiếu.

    datasets[file_name] = pd.read_csv(file_path)  # Đọc file CSV và lưu DataFrame vào dictionary.

for file_name, df in datasets.items():  # Lặp qua toàn bộ dataset vừa đọc.
    print(f'{file_name}: {df.shape}')  # In tên file và kích thước dataset.
```

```python
# ===== CELL 3: KIỂM TRA CHẤT LƯỢNG 5 DATASET =====

for file_name, df in datasets.items():  # Lặp qua từng dataset đã được đọc ở Cell 2.

    print(f'\n===== {file_name} =====')  # In tiêu đề của dataset hiện tại.

    print('Kích thước:', df.shape)  # In số dòng và số cột của dataset.

    print('Số giá trị NaN:', df.isna().sum().sum())  # Đếm tổng số giá trị bị thiếu trong dataset.

    print('Số dòng trùng:', df.duplicated().sum())  # Đếm số dòng bị trùng hoàn toàn trong dataset.

    numeric_df = df.select_dtypes(include=np.number)  # Chọn các cột có dữ liệu dạng số.

    print('Số giá trị Inf:', np.isinf(numeric_df.to_numpy()).sum())  # Đếm tổng số giá trị vô cực trong các cột số.
```

```python
# ===== CELL 4: KIỂM TRA TÊN VÀ KIỂU DỮ LIỆU CÁC CỘT =====

for file_name, df in datasets.items():  # Lặp qua từng dataset để kiểm tra cấu trúc cột.

    print(f'\n===== {file_name} =====')  # In tiêu đề của dataset hiện tại.

    print('Tên các cột:')  # In tiêu đề danh sách cột.

    print(df.columns.tolist())  # In toàn bộ tên cột của dataset.

    print('\nKiểu dữ liệu:')  # In tiêu đề thông tin kiểu dữ liệu.

    print(df.dtypes)  # In kiểu dữ liệu của từng cột.
```

```python
# ===== CELL 5: XÁC ĐỊNH INPUT VÀ TARGET CHO MLP =====

INPUT_COLUMNS = ['ud', 'uq', 'Id', 'Iq', 'nm']  # Khai báo 5 biến đầu vào của MLP; nm là tốc độ cơ thực tế tính bằng rpm.

TARGET_COLUMNS = ['Rs', 'Ld', 'Lq']  # Khai báo 3 tham số PMSM cần MLP xác định.

for file_name, df in datasets.items():  # Lặp qua từng dataset để kiểm tra các biến đầu vào và target.

    print(f'\n===== {file_name} =====')  # In tên dataset đang được kiểm tra.

    print('\nThống kê INPUT:')  # In tiêu đề phần thống kê input.

    print(df[INPUT_COLUMNS].describe().T)  # Tính và hiển thị thống kê của 5 biến đầu vào.

    print('\nThống kê TARGET:')  # In tiêu đề phần thống kê target.

    print(df[TARGET_COLUMNS].describe().T)  # Tính và hiển thị thống kê của Rs, Ld và Lq.
```

```python
# ===== CELL 6: SO SÁNH PHẠM VI THAM SỐ GIỮA 5 DATASET =====

summary_targets = []  # Tạo danh sách để lưu thống kê target của từng dataset.

for file_name, df in datasets.items():  # Lặp qua từng dataset.

    row = {'Dataset': file_name}  # Tạo một dòng thống kê cho dataset hiện tại.

    for target in TARGET_COLUMNS:  # Lặp qua từng tham số Rs, Ld và Lq.

        row[f'{target}_min'] = df[target].min()  # Lưu giá trị nhỏ nhất của tham số.

        row[f'{target}_max'] = df[target].max()  # Lưu giá trị lớn nhất của tham số.

        row[f'{target}_mean'] = df[target].mean()  # Lưu giá trị trung bình của tham số.

        row[f'{target}_std'] = df[target].std()  # Lưu độ lệch chuẩn của tham số.

    summary_targets.append(row)  # Thêm thống kê dataset hiện tại vào danh sách.

target_summary = pd.DataFrame(summary_targets)  # Chuyển danh sách thống kê thành DataFrame.

print(target_summary.to_string(index=False))  # In toàn bộ bảng thống kê target.
```

```python
# ===== CELL 7: SO SÁNH INPUT GIỮA 5 DATASET =====

summary_inputs = []  # Tạo danh sách để lưu thống kê input của từng dataset.

for file_name, df in datasets.items():  # Lặp qua từng dataset.

    row = {'Dataset': file_name}  # Tạo một dòng thống kê cho dataset hiện tại.

    for feature in INPUT_COLUMNS:  # Lặp qua từng biến input của MLP.

        row[f'{feature}_mean'] = df[feature].mean()  # Lưu giá trị trung bình của input.

        row[f'{feature}_std'] = df[feature].std()  # Lưu độ lệch chuẩn của input.

        row[f'{feature}_min'] = df[feature].min()  # Lưu giá trị nhỏ nhất của input.

        row[f'{feature}_max'] = df[feature].max()  # Lưu giá trị lớn nhất của input.

    summary_inputs.append(row)  # Thêm thống kê dataset hiện tại vào danh sách.

input_summary = pd.DataFrame(summary_inputs)  # Chuyển danh sách thống kê thành DataFrame.

print(input_summary.to_string(index=False))  # In bảng thống kê input của 5 dataset.
```

```python
# ===== CELL 8: KIỂM TRA TƯƠNG QUAN GIỮA INPUT VÀ TARGET =====

for file_name, df in datasets.items():  # Lặp qua từng dataset.

    print(f'\n===== {file_name} =====')  # In tên dataset hiện tại.

    correlation = df[INPUT_COLUMNS + TARGET_COLUMNS].corr()  # Tính ma trận tương quan Pearson giữa input và target.

    print(correlation.loc[INPUT_COLUMNS, TARGET_COLUMNS])  # Chỉ hiển thị tương quan giữa input và target.
```

```python
# ===== CELL 9: KIỂM TRA SỰ THAY ĐỔI CỦA Rs THEO THỜI GIAN =====

for file_name, df in datasets.items():  # Lặp qua từng dataset.

    print(f'\n===== {file_name} =====')  # In tên dataset hiện tại.

    print(df[['Time', 'nm', 'Rs', 'Ld', 'Lq']].iloc[::5000].to_string(index=False))  # Lấy một số mẫu cách đều theo thời gian để quan sát sự thay đổi của tham số.
```

```python
# ===== CELL 10: SO SÁNH INPUT GIỮA 5 DATASET TẠI CÙNG THỜI ĐIỂM =====

time_index = 15000  # Chọn một mẫu thời gian để so sánh trực tiếp giữa 5 dataset.

comparison = []  # Tạo danh sách để lưu dữ liệu so sánh.

for file_name, df in datasets.items():  # Lặp qua từng dataset.

    row = {'Dataset': file_name}  # Tạo một dòng dữ liệu cho dataset hiện tại.

    row['Time'] = df.iloc[time_index]['Time']  # Lấy thời gian của mẫu đang được kiểm tra.

    row['ud'] = df.iloc[time_index]['ud']  # Lấy điện áp trục d.

    row['uq'] = df.iloc[time_index]['uq']  # Lấy điện áp trục q.

    row['Id'] = df.iloc[time_index]['Id']  # Lấy dòng điện trục d.

    row['Iq'] = df.iloc[time_index]['Iq']  # Lấy dòng điện trục q.

    row['nm'] = df.iloc[time_index]['nm']  # Lấy tốc độ cơ thực tế tính bằng rpm.

    row['Rs'] = df.iloc[time_index]['Rs']  # Lấy điện trở stator tại thời điểm đó.

    row['Ld'] = df.iloc[time_index]['Ld']  # Lấy điện cảm trục d tại thời điểm đó.

    row['Lq'] = df.iloc[time_index]['Lq']  # Lấy điện cảm trục q tại thời điểm đó.

    comparison.append(row)  # Thêm dòng hiện tại vào danh sách.

comparison_df = pd.DataFrame(comparison)  # Chuyển danh sách thành DataFrame để dễ quan sát.

print(comparison_df.to_string(index=False))  # In bảng so sánh giữa 5 dataset.
```
